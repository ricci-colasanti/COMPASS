```python
import pandas as pd
import csv
```

### Read in the individual survey

```python
df_individual_survey = pd.read_csv('../data/BlockWorld/compleate_artifical_individual_survey.csv')
```

```python
df_individual_survey['ethnicity'].unique()
```

### Get list of attributes

```python
print(df_individual_survey.columns.tolist())
```

```python
attributes =[ 'age', 'education', 'gender', 'ethnicity']
```

### Get the synthetic household data

```python
with open('artificial_synthetic_population.csv', 'r', newline='', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    fieldnames = reader.fieldnames  # Gets the header names directly
    print(fieldnames)
```

### Create a syntheic population of individuals from the synthetic houshold data

```python
synth_population = open("synthetic_population.csv","w")
with open('artificial_synthetic_population.csv', 'r', newline='', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    
    for row in reader:
        hh_id = row['microdata_id']
        area = row['area_id']
        df = df_individual_survey[df_individual_survey['household_id'] == hh_id]
        for index, ind_row in df.iterrows():
            synth_population.write(f"{area},")
            for atrib in attributes:
                synth_population.write(f"{ind_row[atrib]},")
            synth_population.write("\n")

synth_population.close()
        
```
