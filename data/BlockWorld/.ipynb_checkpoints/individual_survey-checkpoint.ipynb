## Create list of individules in the household survey

```python
import pandas as pd
```

### Load survey and full population


```python
df_household_survey = pd.read_csv('artifical_survey.csv')
df_individual = pd.read_csv('artifical_population.csv')
```

### Get a list of all of the households in the survey

```python
household_survey_ids = df_household_survey['hh_id'].unique()
```

### Copy all the individuals from the full population that occure are members of the list of househols in the survey

```python
df_individual_survey = df_individual[df_individual['household_id'].isin(household_survey_ids)]
```

### Create a csv file of all of the individuals in the survey

```python
df_individual_survey.to_csv('artifical_individual_survey.csv', index=False)
```
